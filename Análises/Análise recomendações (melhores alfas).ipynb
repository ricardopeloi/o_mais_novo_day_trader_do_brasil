# Análises

## Verificar ações que mais apareceram: elas cresceram?

```python
bd_recomendacoes_analise_1 = bd_recomendacoes_original.copy()

var_top_5_recomendacoes = bd_recomendacoes_analise_1.groupby("Nome da Empresa").count()[["Ticker"]].sort_values("Ticker", ascending = False).head().index.to_list()
var_top_5_recomendacoes_ticker = bd_recomendacoes_analise_1.groupby(["Nome da Empresa", "Ticker"]).count()[["index"]].sort_values("index", ascending = False).head()#.index.to_list()

# var_top_5_recomendacoes
var_top_5_recomendacoes_ticker = var_top_5_recomendacoes_ticker.reset_index()["Ticker"].to_list()
var_top_5_recomendacoes_ticker
```

```python
# for recomendacao in var_top_5_recomendacoes:
for recomendacao in var_top_5_recomendacoes_ticker:
    # print(recomendacao)
    

    # bd_recomencao = bd_recomendacoes_analise_1.set_index("Ticker").loc[recomendacao].set_index("Data análise")[["Preço HLC"]].copy()
    bd_recomencao = bd_recomendacoes_analise_1.set_index("Ticker").loc[recomendacao].set_index("Data análise")[["Preço HLC"]].copy()
    # display(bd_recomencao)
    
    var_variacao_dias = (bd_recomencao.index.to_list()[-1] - bd_recomencao.index.to_list()[0]).days
    # display(var_variacao_dias)

    var_variacao_valor = bd_recomencao["Preço HLC"].iloc[-1]/bd_recomencao["Preço HLC"].iloc[0] - 1


    # print(recomendacao + " (" + bd_recomendacoes_analise_1[bd_recomendacoes_analise_1["Nome da Empresa"] == recomendacao].iloc[0].loc["Ticker"] + "): " + "{:.0%}".format(var_variacao_valor))
    print(recomendacao + " (" + bd_recomendacoes_analise_1[bd_recomendacoes_analise_1["Ticker"] == recomendacao].iloc[0].loc["Ticker"] + "): " + "{:.0%}".format(var_variacao_valor) + ", ou " + "{:.0%}".format(var_variacao_valor/(var_variacao_dias/365)) + " a.a")

    display(bd_recomencao)
```

Das 5 ações mais recomendadas, 3 cresceram de 40% a 90% em 1 ano. Teriam sido grandes investimentos.

Já as outras duas ações ficaram no zero a zero, caso o jogador saia no momento certo. São ações bem voláteis, flexionando preço fortemente ao longo dos períodos e podendo gerar perdas potenciais maiores

## Colocando 1000 em cada uma na primeira recomendação até período mais recente

```python
bd_dados_historico = pd.read_excel(var_caminho + r"\Bases\Base de Dados Histórico.xlsx")
bd_dados_historico_top_recomendacoes = bd_dados_historico[
    (bd_dados_historico["Date"] == max(bd_dados_historico["Date"]))
    *
    (bd_dados_historico["Ticker"].isin(var_top_5_recomendacoes_ticker))
][["Ticker", "Date", "HLC"]]

bd_dados_historico_top_recomendacoes
```

```python
var_data_maxima = max(bd_dados_historico_top_recomendacoes["Date"])

# for recomendacao in var_top_5_recomendacoes:
for recomendacao in var_top_5_recomendacoes_ticker:
    print(recomendacao)
    

    # bd_recomencao = bd_recomendacoes_analise_1.set_index("Ticker").loc[recomendacao].set_index("Data análise")[["Preço HLC"]].copy()
    bd_recomencao = bd_recomendacoes_analise_1.set_index("Ticker").loc[recomendacao].set_index("Data análise")[["Preço HLC"]].copy()
    # display(bd_recomencao)
    
    var_variacao_dias = (var_data_maxima - bd_recomencao.index.to_list()[0]).days
    # display(var_variacao_dias)

    var_variacao_valor = bd_dados_historico_top_recomendacoes.set_index("Ticker").loc[recomendacao, "HLC"]/bd_recomencao["Preço HLC"].iloc[0] - 1
    print("Preço inicial: " + "{:.2f}".format(bd_recomencao["Preço HLC"].iloc[0]))
    print("Preço final: " + "{:.2f}".format(bd_dados_historico_top_recomendacoes.set_index("Ticker").loc[recomendacao, "HLC"]))


    # print(recomendacao + " (" + bd_recomendacoes_analise_1[bd_recomendacoes_analise_1["Nome da Empresa"] == recomendacao].iloc[0].loc["Ticker"] + "): " + "{:.0%}".format(var_variacao_valor))
    print("Crescimento: " + "{:.0%}".format(var_variacao_valor) + ", ou " + "{:.0%}".format(var_variacao_valor/(var_variacao_dias/365)) + " a.a")

    # display(bd_recomencao)
    print()
```
